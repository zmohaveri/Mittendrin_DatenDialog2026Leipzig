## Direkt API ansprechen -> Website digitalpakt-alter.de

```python
import requests
import json
```

```python
url = "https://www.digitalpakt-alter.de/"
params = {
    "type": "1698763347",
    "pidList": "2163,1268,2328",
    "id": "1272",
}
headers = {
    "accept": "*/*",
    "accept-language": "de,de-DE;q=0.9,en;q=0.8",
    "referer": "https://www.digitalpakt-alter.de/angebote-finden/",
    "user-agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
        "(KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"
    ),
}
cookies = {"CookieConsent": "mandatory"}
```

```python
resp = requests.get(url, params=params, headers=headers, cookies=cookies, timeout=15)

print("Status:", resp.status_code)
print("Content-Type:", resp.headers.get("content-type"))
print("Erste 2000 Zeichen der Antwort:\n")
print(resp.text[:2000])
```

```python
with open("digitalpakt_alter_angebote.json", "w", encoding="utf-8") as f:
    f.write(resp.text)

print("\n--> Komplette Antwort in angebote_raw.json gespeichert.")
```

### Ergebnis anschauen

```python
# --- In DataFrame laden ---
import pandas as pd
```

```python
data = json.loads(resp.text)
records = data["data"]["stations"]
 
df = pd.json_normalize(records)
df["zip"] = df["zip"].astype(str)
print(df.shape)
 
# Brandenburg über PLZ-Präfixe filtern (Brandenburg: 01, 03, 04, 12-19 je nach Grenzregion,
# grob abgedeckt durch 12xxx-19xxx + 03xxx; ggf. nachschärfen anhand city-Spalte)
bb_prefixes = ("03", "14", "15", "16")
df_bb = df[df["zip"].str.startswith(bb_prefixes)]
print(df_bb.shape)
df_bb.head(20)
```

```python
df_bb.to_json("digitalpakt_alter_angebote_restructured.json", orient="records", force_ascii=False, indent=2)
```
